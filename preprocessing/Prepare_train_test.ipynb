```python
import pandas as pd 

# Save to CSV
train_file = "../data/train_proteins.tsv"
val_file = "../data/val_proteins.tsv"
test_file = "../data/test_proteins.tsv"
```

```python
import pandas as pd

# Load your dataframe
df = pd.read_csv(train_file,sep='\t')

# Columns you want to keep
columns_to_keep = [
    "Entry", "Gene Ontology IDs", "Function [CC]"
]

# Filter the DataFrame
filtered_df = df[columns_to_keep]

# Save to CSV if needed
filtered_df.to_csv("../data/train_dataset.tsv",sep='\t', index=False)
```
