#### This is the script to make GO Document 

```python
import os
import re
import random

def parse_go_obo(obo_file_path):
    go_dict = {}
    with open(obo_file_path, 'r', encoding='utf-8') as f:
        current_go = None
        current_def = None

        for line in f:
            if line.startswith("[Term]"):
                current_go = None
                current_def = None

            elif line.startswith("id: GO:"):
                current_go = line.strip().split("id: ")[1]

            elif line.startswith("def:"):
                match = re.search(r'"(.*?)"', line)
                if match:
                    current_def = match.group(1)

            elif line.strip() == "" and current_go and current_def:
                go_dict[current_go] = current_def

    return go_dict

def remove_evidence_code(text):
    text = re.sub(r'\[.*?\]', '', text)
    text = re.sub(r'\[|\]', '', text)
    return text.strip()

def get_go_description_from_obo(go_dict, goterms, randomize=True):
    if randomize:
        random.shuffle(goterms)

    descriptions = []
    for goterm in goterms:
        if goterm in go_dict:
            desc = remove_evidence_code(go_dict[goterm])
            if not desc.endswith('.'):
                desc += '.'
            descriptions.append(desc)
        else:
            print(f"Warning: GO term {goterm} not found in GO dictionary.")

    return ' '.join(descriptions)

# Example usage:
if __name__ == "__main__":
    obo_path = "../data/go-basic.obo"
    goterms = ["GO:0006915", "GO:0008150", "GO:0009987"]  # example GO terms

    go_data = parse_go_obo(obo_path)
    go_doc = get_go_description_from_obo(go_data, goterms)
    print(go_doc)
```
