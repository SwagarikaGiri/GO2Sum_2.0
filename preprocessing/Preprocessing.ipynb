## This script will generate the training testing dataset
### The filteration that we need to do is check which proteins dont have GO term annotated
### Then randomly sample 80/10/10  for train, dev, test

```python
import pandas as pd 


uniprot_file_name= '../data/uniprotkb_reviewed_true_AND_model_organ_2025_05_08.tsv'
df = pd.read_csv(uniprot_file_name,sep='\t')
print(df.head(5))
```

```python
# Filter rows where "Gene Ontology IDs" contains "GO:"
filtered_df = df[df["Gene Ontology IDs"].str.contains("GO:", na=False)]

# Save the filtered dataframe
filtered_df.to_csv("../data/human_proteins_dataset.tsv",sep='\t', index=False)
```

```python
# Shuffle the filtered dataframe
filtered_df = filtered_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Compute sizes
total = len(filtered_df)
train_size = int(total * 0.8)
val_size = int(total * 0.1)
test_size = total - train_size - val_size  # Remaining

# Split
train_df = filtered_df.iloc[:train_size]
val_df = filtered_df.iloc[train_size:train_size + val_size]
test_df = filtered_df.iloc[train_size + val_size:]

# Save to CSV
train_df.to_csv("../data/train_proteins.tsv",sep='\t', index=False)
val_df.to_csv("../data/val_proteins.tsv", sep='\t',index=False)
test_df.to_csv("../data/test_proteins.tsv",sep='\t', index=False)

print("Done! Train:", len(train_df), "Val:", len(val_df), "Test:", len(test_df))
```

```python
import pandas as pd
df = pd.read_csv("../data/train_dataset.tsv", sep="\t")
df = df.dropna(subset=["Gene Ontology IDs", "Function [CC]"])
df = df[df["Function [CC]"].str.strip() != ""]
df.to_csv("../data/train_dataset_cleaned.tsv", sep="\t", index=False)

df = pd.read_csv("../data/val_dataset.tsv", sep="\t")
df = df.dropna(subset=["Gene Ontology IDs", "Function [CC]"])
df = df[df["Function [CC]"].str.strip() != ""]
df.to_csv("../data/val_dataset_cleaned.tsv", sep="\t", index=False)

df = pd.read_csv("../data/test_dataset.tsv", sep="\t")
df = df.dropna(subset=["Gene Ontology IDs", "Function [CC]"])
df = df[df["Function [CC]"].str.strip() != ""]
df.to_csv("../data/test_dataset_cleaned.tsv", sep="\t", index=False)
```
